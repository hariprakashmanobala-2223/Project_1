# Stock Price Prediction Using Machine Learning

Starter notebook for exploring historical stock closing prices. Place a permitted CSV dataset at `data/stock_data.csv` with `Date` and `Close` columns.

**Disclaimer:** Educational demonstration only; not financial advice.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

data_path = Path('../data/stock_data.csv')
if not data_path.exists():
    raise FileNotFoundError('Add data/stock_data.csv first. See the project README.')
df = pd.read_csv(data_path, parse_dates=['Date']).sort_values('Date')
df.head()
```

```python
required = {'Date', 'Close'}
missing = required - set(df.columns)
if missing:
    raise ValueError(f'Missing required columns: {missing}')
df = df.dropna(subset=['Date', 'Close'])
df.plot(x='Date', y='Close', figsize=(10, 4), title='Historical Closing Price', legend=False)
plt.ylabel('Closing price')
plt.tight_layout()
plt.show()
```
